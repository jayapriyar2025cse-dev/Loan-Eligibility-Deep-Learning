```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input,Dense
```

```python
df=pd.read_csv(r"/content/Loan-Approval-Prediction.csv")
print(df.head())
```

```python
# Handle missing values

# Impute categorical features with their mode
for col in ['Gender', 'Married', 'Dependents', 'Self_Employed', 'Credit_History']:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# Impute numerical features with their median
for col in ['LoanAmount', 'Loan_Amount_Term']:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

# Verify that there are no more missing values
print("Missing values after imputation:")
print(df.isnull().sum())
```

```python
x = df.drop("Loan_Status", axis=1)
y = df["Loan_Status"]

# Drop Loan_ID as it's an identifier and not a feature
x = x.drop("Loan_ID", axis=1)

# Convert categorical features to numerical using one-hot encoding
x = pd.get_dummies(x, drop_first=True)

# Convert target variable y to numerical (0 and 1)
y = y.map({'Y': 1, 'N': 0})
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
#scale the data
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
#create neural network
model=Sequential([
    #input->hidden layer
    #12 input features
    #ReLU activation
  Input(shape=(x_train.shape[1],)),
    Dense(12, activation="relu"),
    #hidden layer->output layer
    #1 o/p neuron
    #sigmoid gives probability b/w 0 and 1
    Dense(1,activation="sigmoid")
])
```

```python
#compile the model
model.compile(#adam
              #adam will update the weight during training
              #it uses gradients calculated by backprobagation
              optimizer="adam",
              #loss
              #measure how wrong the prediction is
              loss = "binary_crossentropy",
              metrics=["accuracy"])
```

```python
# Test dataset accuracy
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
```

```python
model.fit(x_train,y_train,
          #repeat learning 20 times
          epochs=20,
          #process 16 records at a time
          batch_size=16,
          verbose=1)
```

```python
prediction=model.predict(x_test[:1],verbose=0)
#get probability value
probability=prediction[0][0]
print("\n Prediction Probability : ",probability)
if(probability>=0.5):
  print("Predicted class :1")
  print("Meaning : Loan approved")
else:
  print("Predicted class :0")
  print("Meaning : loan not approved")
```
